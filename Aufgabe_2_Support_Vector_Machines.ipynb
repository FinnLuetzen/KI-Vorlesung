```python
import pandas as pd
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
from google.colab import files

import warnings
warnings.filterwarnings('ignore')
```

Datensatz laden und vorbereiten (nicht bearbeiten):

```python
train_url = 'https://raw.githubusercontent.com/TNemes-3141/kiml-gn-public-datasets/refs/heads/main/task%202/train.csv'
test_url = 'https://raw.githubusercontent.com/TNemes-3141/kiml-gn-public-datasets/refs/heads/main/task%202/test.csv'

train_df = pd.read_csv(train_url)
test_df = pd.read_csv(test_url)

print("Trainingsdatensatz (Ausschnitt):")
print(train_df.head())
print("\nTestdatensatz (Ausschnitt):")
print(test_df.head())

# Features und Labels trennen
X_train_raw = train_df.drop(columns=['label'])
y_train = train_df['label'].values
X_test_raw = test_df.copy()

# Daten skalieren (Standardisierung)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train_raw)
X_test_scaled = scaler.transform(X_test_raw)

print("\nDaten erfolgreich geladen und vorbereitet!")
```
